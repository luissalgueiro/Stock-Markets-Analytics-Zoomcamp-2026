````python

# Web Scraping for Macro
import requests
import io
from bs4 import BeautifulSoup
url = "https://en.wikipedia.org/wiki/List_of_S%26P_500_companies"
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/58.0.3029.110 Safari/537.3"
}
response = requests.get(url, headers=headers)
io_stream = io.StringIO(response.text)
# Read tables from HTML string
tables = pd.read_html(io_stream)


print(tables)
````

## STEPS

1. Create a DataFrame with company tickers, names, and the year they were added

````python
df = tables[0][['Symbol','Security','Date added']]
df.head()
````

2. Extract the year from the addition date and calculate the number of stocks added each year.

````python
df['Year'] = pd.to_datetime(df['Date added'], format='mixed').dt.year
df.head()
````

````python
stocks_per_year = df.groupby(by='Year').size().reset_index().rename(columns={0:'stocks_per_year'})
stocks_per_year.head()
````

3. Which (full) year had the highest number of additions, starting from 2020?


````python
full_year_max_from_2020 = stocks_per_year[stocks_per_year['Year']>2019].max()

print(f"The full year that had the higgest number of additions, starting from 2020 was: {full_year_max_from_2020['Year']} with {full_year_max_from_2020['stocks_per_year']} stocks")
````

Additional: How many current S&P 500 stocks have been in the index for more than 20 years?



````python
df['date_added_dt'] = pd.to_datetime(df['Date added'], format='mixed', errors='coerce')
today  = pd.Timestamp.now()

df['current_years'] = (today-df['date_added_dt'])/pd.Timedelta(days=365)
more_than_20y = df[df['current_years']>20]
display(more_than_20y.head())
print(f"Total de empresas con más de 20 años en el índice: {len(more_than_20y)}")
````

## Question 2:

Indexes YTD (as of 21 August 2026)
How many indexes (out of 10) have better year-to-date returns than the US (S&P 500) as of August 21, 2026?

- United States - S&P 500 (^GSPC)
- China - Shanghai Composite (000001.SS)
- Hong Kong - HANG SENG INDEX (^HSI)
- Australia - S&P/ASX 200 (^AXJO)
- India - Nifty 50 (^NSEI)
- Canada - S&P/TSX Composite (^GSPTSE)
- Germany - DAX (^GDAXI)
- United Kingdom - FTSE 100 (^FTSE)
- Japan - Nikkei 225 (^N225)
- Mexico - IPC Mexico (^MXX)
- Brazil - Ibovespa (^BVSP)



````python
import yfinance as yf
````

````python
df_full = pd.DataFrame(columns=['Ticker','YTD']) 

tickers  = ['^GSPC','000001.SS','^HSI','^AXJO','^NSEI','^GSPTSE','^GDAXI','^FTSE','^N225','^MXX','^BVSP']
for ticker in tickers:
    # Download historical data (e.g., for the year 2026 up to August 21)
    data = yf.download(ticker, start="2026-01-01", end="2026-08-21")
    # Display the first few rows
    # display(data.head())
    ytd = ((data.iloc[-1][('Close', ticker)] - data.iloc[0][('Close', ticker)])/ data.iloc[0][('Close', ticker)])*100
    print(f'Company {ticker} has a {ytd:.2f}% return')
    df_dict  = pd.DataFrame({'Ticker':ticker,'YTD':ytd},index=[0])
    df_full = pd.concat((df_full, df_dict)).reset_index(drop=True)



    # break
````

````python
df_full['Rate'] = (df_full['YTD'] / df_full['YTD'].iloc[0]) * 100
````

````python
print(f"There are {df_full[df_full['Rate']>100].shape[0]} indexes with better YTD than SP500")
````
