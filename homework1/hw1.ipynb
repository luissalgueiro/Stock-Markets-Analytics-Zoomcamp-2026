````python

# Web Scraping for Macro
import requests
import io
from bs4 import BeautifulSoup
url = "https://en.wikipedia.org/wiki/List_of_S%26P_500_companies"
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/58.0.3029.110 Safari/537.3"
}
response = requests.get(url, headers=headers)
io_stream = io.StringIO(response.text)
# Read tables from HTML string
tables = pd.read_html(io_stream)


print(tables)
````

## STEPS

1. Create a DataFrame with company tickers, names, and the year they were added

````python
df = tables[0][['Symbol','Security','Date added']]
df.head()
````

2. Extract the year from the addition date and calculate the number of stocks added each year.

````python
df['Year'] = pd.to_datetime(df['Date added'], format='mixed').dt.year
df.head()
````

````python
stocks_per_year = df.groupby(by='Year').size().reset_index().rename(columns={0:'stocks_per_year'})
stocks_per_year.head()
````
